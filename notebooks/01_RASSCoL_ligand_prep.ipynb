# Ligand preparation

Code to prepare ligands of interest for the RASSCoL pipeline. If you do not have a structure file, this notebook can be used to create one from a SMILES.

**ALWAYS VISUALLY INSPECT THE FILES GENERATED FROM SMILES**

## Directory creation

```python
# builtins
from pathlib import Path
import sys

sys.path.append('..')

# local
from rasscol_src.general_utils import *
from rasscol_src.rasscol_utils import *

## USER INPUT ##
data_dir = Path('../example/scapCC4_NRD')
ligand_name = 'nile_red'
ligand_short_name = 'NRD' # must be three alphanumeric characters only

# pdb_path = Path('example.pdb') # or provide a SMILES if not available
smiles = 'CCN(CC)C1=CC2=C(C=C1)N=C3C4=CC=CC=C4C(=O)C=C3O2'
#################
```

## SMILES to PDB

If you already have a structure file ignore this cell.

```python
# Ensure the target directory for storing data exists; create it if it doesn't.
data_dir.mkdir(parents=True, exist_ok=True)

# Convert the SMILES string to an RDKit molecule and add hydrogens to the molecule.
mol = smiles2mol(smiles, addH=False)

# Minimize the energy of the molecule to obtain a more stable conformation.
mol_minimized = minimize_mol(mol)

# Save the minimized molecule to a PDB file and get the path to the saved file.
pdb_path = mol2pdb(ligand_name, mol_minimized, data_dir)
```

## PDB to PDBQT

```python
# Convert the PDB file to a PDBQT file using Open Babel, preparing it for docking.
ligand_pdbqt_path = pdb2pdbqt(pdb_path)

# Rename the residue in the PDBQT file from "UNL" to the short name provided
tidy_ligand_pdbqt(ligand_pdbqt_path, ligand_short_name)
```

## Ligand volume/length calculation

```python

# Extract the coordinates of the ligand from the PDBQT file.
ligand_coords = get_pdbqt_coords(ligand_pdbqt_path)

# Calculate the length of the molecule based on its coordinates.
mol_length = get_mol_len(ligand_coords)

# Calculate the Bondi volume of the molecule based on its molecular formula and ring counts.
bondi_vol = calc_bondi_vol_from_mol(mol)

# Print the calculated Bondi volume in cubic angstroms (Å³) and molecule length (Å).
print(f"Bondi volume: {bondi_vol} Å3")
print(f"Length: {mol_length:.1f} Å")
```
